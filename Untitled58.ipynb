```python
import pandas as pd
from IPython.display import display

# 1. 새로운 파일 불러오기 및 기본 세팅
df = pd.read_excel('kbank.xlsx', header=3)
df = df.dropna(subset=['거래일시'])
df['출금금액'] = pd.to_numeric(df['출금금액'], errors='coerce').fillna(0)

# (중요) 날짜 필터링을 위해 파이썬에게 '이거 날짜야!'라고 알려주기
df['거래일시'] = pd.to_datetime(df['거래일시'])

# 2. 👑 영석님의 기획 조건 3가지 동시 타격! (if문 대신 조건 필터링 사용)
# 분석하고 싶은 달(예: 8월)을 지정합니다. (9월을 원하시면 8을 9로 바꾸면 됩니다)
target_month = 8

cond1 = df['출금금액'] > 0                   # 기획 1: 0원짜리 껍데기 결제는 빼라!
cond2 = df['거래구분'] != '플러스박스이체'     # 기획 2: 플러스박스는 지출이 아니니 빼라!
cond3 = df['거래일시'].dt.month == target_month # 기획 3: 딱 한 달(8월)치만 가져와라!

# 3가지 조건을 모두 만족(&)하는 진짜 알짜배기 데이터만 복사
real_spend = df[cond1 & cond2 & cond3].copy()
clean_df = real_spend[['거래일시', '거래구분', '출금금액', '상대 예금주명', '적요내용']]

# 3. 10개 제한 해제! 엑셀처럼 전체 데이터 쫙 뿌리기
print(f"📅 [{target_month}월 한 달] 0원 제외, 진짜 지출 전체 내역")
# head(10)을 지워버려서 전체 데이터가 다 나옵니다!
display(clean_df)

# 4. 기획 4: "어디에 돈을 제일 많이 썼을까?" (그룹 묶어서 랭킹 매기기)
# '상대 예금주명' 끼리 묶어서 출금금액을 다 더한 뒤, 비싼 순서대로 줄 세우기!
top_spending = clean_df.groupby('상대 예금주명')['출금금액'].sum().sort_values(ascending=False)

print(f"\n🏆 {target_month}월 지출 랭킹 Top 5 (제일 돈 많이 쓴 곳)")
display(top_spending.head(5).to_frame(name='총 쓴 돈(원)'))

# 5. 한 달 치 최종 합계
total_expense = clean_df['출금금액'].sum()
print(f"\n🎉 {target_month}월 진짜 지출 총합: {total_expense:,.0f} 원")
```

```python
import pandas as pd
from IPython.display import display

# 1. 새로운 파일 불러오기 및 찌꺼기 정제
df = pd.read_excel('kbank.xlsx', header=3)
df = df.dropna(subset=['거래일시'])
df['출금금액'] = pd.to_numeric(df['출금금액'], errors='coerce').fillna(0)
df['거래일시'] = pd.to_datetime(df['거래일시'])

# 2. 전체 데이터에서 0원 결제와 플러스박스 걷어내기 (영석님의 비즈니스 로직!)
cond1 = df['출금금액'] > 0
cond2 = df['거래구분'] != '플러스박스이체'
real_spend = df[cond1 & cond2].copy()

# 달(Month) 정보만 쏙 뽑아서 새로운 기둥(컬럼) 만들기
real_spend['월'] = real_spend['거래일시'].dt.month

# 3. 영석님의 기획: "7월, 8월, 9월 각각 총합이랑 Top 4를 뽑아라!"
months_to_check = [7, 8, 9]

for month in months_to_check:
    # 해당 월(month)의 데이터만 쏙 필터링
    monthly_data = real_spend[real_spend['월'] == month]

    # 만약 그 달에 쓴 내역이 없다면 그냥 넘어가기
    if monthly_data.empty:
        continue

    # 그 달의 총 지출액 계산
    monthly_total = monthly_data['출금금액'].sum()

    # 출력 꾸미기 (보고서 느낌 팍팍 나게!)
    print("=" * 50)
    print(f" 📅 [{month}월] 총 지출 금액 : {monthly_total:,.0f} 원")
    print("=" * 50)

    # 그 달의 지출 랭킹 Top 4 뽑기
    top4 = monthly_data.groupby('상대 예금주명')['출금금액'].sum().sort_values(ascending=False).head(4)

    print(f" 🏆 {month}월 내 돈을 가장 많이 가져간 곳 Top 4")
    display(top4.to_frame(name='총 쓴 돈(원)'))
    print("\n") # 다음 달로 넘어가기 전에 보기 좋게 한 줄 띄우기
```
